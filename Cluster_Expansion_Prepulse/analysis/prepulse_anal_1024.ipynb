```python
import numpy as np
import scipy.constants as const
import matplotlib.pyplot as plt


lambda0 = 0.8e-6
c = const.c

omega0 = 2*np.pi*c/lambda0
Tlaser = 2*np.pi/omega0

# acceleration time = half a laser period
dt_acc = Tlaser/2


def electron_energy(Efield):

    p = const.e * Efield * dt_acc

    gamma = np.sqrt(
        1 + (p/(const.m_e*c))**2
    )

    KE = (gamma-1)*const.m_e*c**2

    return KE/const.e   # eV


E0 = 1e12

print("Laser period =", Tlaser*1e15, "fs")
print("Acceleration time =", dt_acc*1e15, "fs")
print("Electron energy =", electron_energy(E0), "eV")
```

```python
Te = 2/3 * electron_energy(E0)

print("Temperature =", Te, "eV")
```

```python
T = S.TrackParticles("hydrogen_clusters")

times = T.getAvailableTimesteps()

density = []
time_ps = []

for t in times:

    data = T.getData(timestep=t)

    x = data["x"]
    y = data["y"]

    if "w" in data:
        N = np.sum(data["w"])
    else:
        N = len(x)

    # RMS widths
    sigma_x = np.std(x)
    sigma_y = np.std(y)

    # FWHM widths
    fwhm_x = 2.355*sigma_x
    fwhm_y = 2.355*sigma_y

    # Approximate cluster area (2D simulation)
    area = np.pi*(fwhm_x/2)*(fwhm_y/2)

    # Number density (2D estimate)
    n = N/area

    density.append(n)

    time_ps.append(dt*t/omega0*1e12)

print(density)
plt.figure(figsize=(7,4))
plt.plot(time_ps, density, lw=2)

plt.xlabel("Time (ps)")
plt.ylabel(r"Estimated cluster density ($n/n_c$)")
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
T = S.TrackParticles("carbon_clusters")

t_final = T.getAvailableTimesteps()[0]

data = T.getData(timestep=t_final)

x = np.array(data["x"])
y = np.array(data["y"])

mask = np.isfinite(x) & np.isfinite(y)

x = x[mask]
y = y[mask]

print("Remaining particles:", len(x))


def compute_fwhm(coords, bins=200):
    """
    Compute FWHM of a 1D particle distribution using the histogram
    half-maximum method.
    """
    hist, edges = np.histogram(coords, bins=bins)
    centres = 0.5 * (edges[:-1] + edges[1:])

    half_max = hist.max() / 2

    above = np.where(hist >= half_max)[0]

    if len(above) < 2:
        return np.nan, centres, hist, half_max

    left = centres[above[0]]
    right = centres[above[-1]]

    fwhm = right - left

    return fwhm, centres, hist, half_max


# Compute FWHM in x and y
fwhm_x, centres_x, hist_x, hm_x = compute_fwhm(x)
fwhm_y, centres_y, hist_y, hm_y = compute_fwhm(y)

lambda0 = 800e-9  # m

fwhm_m_x = fwhm_x * lambda0 / (2*np.pi)
fwhm_nm_x = fwhm_m_x * 1e9

fwhm_m_y = fwhm_y * lambda0 / (2*np.pi)
fwhm_nm_y = fwhm_m_y * 1e9

print(f"Final timestep = {t_final}")
print(f"FWHM_x = {fwhm_nm_x:.6f} nm")
print(f"FWHM_y = {fwhm_nm_y:.6f} nm")

# Plot distributions
fig, ax = plt.subplots(1, 2, figsize=(12,4))

ax[0].plot(centres_x, hist_x)
ax[0].axhline(hm_x, color='r', linestyle='--')
ax[0].set_xlabel("x")
ax[0].set_ylabel("Particle count")
ax[0].set_title(f"x distribution\nFWHM = {fwhm_x:.4f}")

ax[1].plot(centres_y, hist_y)
ax[1].axhline(hm_y, color='r', linestyle='--')
ax[1].set_xlabel("y")
ax[1].set_ylabel("Particle count")
ax[1].set_title(f"y distribution\nFWHM = {fwhm_y:.4f}")

plt.tight_layout()
plt.show()
```

```python
C = S.TrackParticles("carbon_clusters")

timesteps = C.getTimesteps()

n0 = 91.0/5      
R0_nm = 10.0

Lx = 1e-6/c_over_omega0
Ly = Lx

xc = 0.5 * Lx
yc = 0.5 * Ly

density_nc = []
radius_nm = []

for t in timesteps:

    data = C.getData(timestep=t)

    x = np.array(data["x"])
    y = np.array(data["y"])

    r = np.sqrt((x - xc)**2 + (y - yc)**2)

    r_nm = r * c_over_omega0 * 1e9

    R = np.percentile(r_nm, 90) # avoids escaped ions from dominating

    radius_nm.append(R)

    # 2D expansion: density scales as 1/R^2
    n = n0 * (R0_nm / R)**2

    density_nc.append(n)

radius_nm = np.array(radius_nm)
density_nc = np.array(density_nc)

# Convert timesteps to ps
times = np.array(timesteps) * dt / omega0 * 1e12
```

```python
plt.figure(figsize=(7,4))

plt.plot(times, radius_nm)

plt.xlabel("Time [ps]")
plt.ylabel("90th percentile ion radius [nm]")
plt.title("Carbon cluster expansion")

plt.grid()
plt.show()
```

```python
plt.figure(figsize=(7,4))

plt.plot(times, density_nc)

plt.axhline(5, linestyle="--", label="$5n_c$")

plt.xlabel("Time [ps]")
plt.ylabel(r"Estimated density [$n_c$]")
plt.title("Estimated carbon ion density")
plt.xlim(0.4,0.6)
#plt.ylim(0.01, 1000)
plt.yscale("log")
plt.grid()
plt.legend()
plt.savefig('carbon_density_evol.pdf', dpi=400, bbox_inches='tight')
plt.show()
```
